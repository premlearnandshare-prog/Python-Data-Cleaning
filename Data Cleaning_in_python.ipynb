```python
# Car_name 
x = df['car_name'][0]
x[:x.index(' '):1]
```

```python
def company_name(x):
    return x[:x.index(' '):1]
```

```python
df['company_name'] = df['car_name'].apply(company_name)
```

```python
df['car_name'][0]
```

```python
def car_name(x):
    return x[x.index(' ') + 1::1]
```

```python
df['car_name'] = df['car_name'].apply(car_name)
```

```python
df['car_name']
```

```python
# car_price_in_rupees
```

```python
df['car_price_in_rupees'] = df['car_price_in_rupees'].str.replace("," , "")
```

```python
df['car_price_in_rupees'][0]
```

```python
def rupe_change(x):
    p = x.split(' ')
    try :
        if p[1] == 'Lakh' :
            return str(round(float(p[0])*100000,1))
        elif p[1] == 'Crore' :
            return str(round(float(p[0])*10000000, 1))
    except :
        return x
```

```python
df['car_price_in_rupees'] = df['car_price_in_rupees'].apply(rupe_change)
```

```python
df['car_price_in_rupees']
```

```python
df['car_price_in_rupees'].dtype
```

```python
df['car_price_in_rupees'] = df['car_price_in_rupees'].astype('float')
```

```python
df['kms_driven'] = df['kms_driven'].str.replace(",","")
```

```python
def kms_change(x):
    return x[:x.index(' '):]
```

```python
df['kms_driven'] = df['kms_driven'].apply(kms_change)
```
